# 07 - Results Visualization

Portfolio-level dashboards and export-ready figures using RiskCharts.

## Objectives
- Use RiskCharts for all chart types
- Build portfolio-level dashboards
- Create export-ready figures

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path('..').resolve()))

import pandas as pd
from src.visualization.risk_charts import RiskCharts
```

## 1. Load Processed Data with Rankings

```python
df = pd.read_csv('../data/processed/projects_with_features.csv') if Path('../data/processed/projects_with_features.csv').exists() else pd.read_csv('../data/raw/sample_projects.csv')

if 'risk_level' not in df.columns:
    if 'mcda_score' in df.columns:
        df['risk_level'] = df['mcda_score'].apply(lambda x: 'Low' if x >= 0.7 else ('Medium' if x >= 0.4 else 'High'))
    elif 'risk_score' in df.columns:
        df['risk_level'] = df['risk_score'].apply(lambda x: 'High' if x >= 0.6 else ('Medium' if x >= 0.3 else 'Low'))

print(f'Loaded {len(df)} projects')
```

## 2. Risk Distribution Pie Chart

```python
if 'risk_level' in df.columns:
    fig = RiskCharts.risk_distribution_pie(df)
    fig.show()
    fig.write_html('risk_distribution.html')
    print('Saved risk_distribution.html')
else:
    print('Run rankings first to get risk_level')
```

## 3. Risk Score Bar Chart

```python
score_col = 'mcda_score' if 'mcda_score' in df.columns else 'risk_score'
if score_col in df.columns:
    fig = RiskCharts.risk_score_bar(df, top_n=10, score_col=score_col)
    fig.show()
    fig.write_html('risk_score_bar.html')
else:
    print('No risk score column')
```

## 3b. Feature Importance

```python
import pandas as pd
imp_df = pd.DataFrame({'feature': ['completion_rate', 'budget_variance', 'schedule_performance', 'team_size'], 'importance': [0.25, 0.20, 0.18, 0.15]})
fig = RiskCharts.feature_importance_bar(imp_df, top_n=10)
fig.show()
```

## 4. Risk Gauge and Radar Charts

```python
fig = RiskCharts.risk_gauge(0.35, title='Portfolio Risk Score')
fig.show()

categories = {'technical': 5, 'resource': 3, 'schedule': 7, 'scope': 2, 'budget': 4}
fig2 = RiskCharts.risk_category_radar(categories)
fig2.show()
```
